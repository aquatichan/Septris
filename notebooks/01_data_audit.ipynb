## Variable availability

Two drop rules are shown side by side:
- `flag_hours_rule`: recorded in fewer than 1% of patient-hours at either hospital (stricter; also removes labs rare at both).
- `flag_hospital_specific`: recorded for fewer than 1% of patients at one hospital but not the other (the actual hidden-hospital risk).

The team picks one before any cross-hospital modelling.

```python
av = pd.read_csv('outputs/availability.csv')
av.sort_values('A_hours_pct')
```

```python
print('Hours rule:', av.loc[av.flag_hours_rule, 'variable'].tolist())
print('Hospital-specific rule:', av.loc[av.flag_hospital_specific, 'variable'].tolist())
```
